# Exploratory Data Analysis (EDA): Music Context & Graphs
This notebook analyzes the audio features, chord progressions, text tag distributions, and graph topology.

```python
import os
import sys
import json
import numpy as np
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath("../src"))

metadata_path = "../data/raw/dataset_metadata.json"
if os.path.exists(metadata_path):
    with open(metadata_path, "r") as f:
        meta = json.load(f)
    print(f"Total tracks: {len(meta['tracks'])}")
    print(f"Tag vocabulary ({len(meta['tags_vocab'])}):", meta["tags_vocab"])
else:
    print("Run sample_data_gen.py to generate sample metadata first.")
```

## 1. Tag and Valence / Arousal Distributions

```python
if os.path.exists(metadata_path):
    valences = [t["valence"] for t in meta["tracks"]]
    arousals = [t["arousal"] for t in meta["tracks"]]
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
    ax1.hist(valences, bins=10, color='skyblue', edgecolor='black')
    ax1.set_title("DEAM Valence Distribution")
    ax1.set_xlabel("Valence (1-9)")
    
    ax2.hist(arousals, bins=10, color='salmon', edgecolor='black')
    ax2.set_title("DEAM Arousal Distribution")
    ax2.set_xlabel("Arousal (1-9)")
    plt.tight_layout()
    plt.show()
```

## 2. Graph Connectivity & Degree Statistics
